# GANs

**GAN (Generative Adversarial Network)** has two neural networks playing a game:
* Generator tries to create realistic fake images
* Discriminator tries to detect if an image is real or fake

They train together:
* Generator gets better at fooling Discriminator
* Discriminator gets better at spotting fakes

Eventually, the generator learns to generate realistic-looking images

```python
# Importing Libraries
import numpy as np
import matplotlib.pyplot

from tensorflow.keras import Model, Input
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (Dense, Reshape, Conv2DTranspose, 
    Conv2D, LeakyReLU, Flatten, Dropout, BatchNormalization)

# 1. Load and preprocess data:
(X_train, _), _ = cifar10.load_data()
X_train = (X_train.astype(np.float32) - 127.5) / 127.5  # [-1, 1]
latent_dim = 100
```

```python
# 2. Build the Generator (ConvTranspose Network):
def build_generator(latent_dim=100):
    model = Sequential(name="generator")
    model.add(Dense(256 * 4 * 4, input_dim=latent_dim))
    model.add(LeakyReLU(0.2))
    model.add(Reshape((4, 4, 256)))                      # 4x4

    model.add(Conv2DTranspose(128, kernel_size=4, strides=2, padding="same"))  # 8x8
    model.add(BatchNormalization())
    model.add(LeakyReLU(0.2))

    model.add(Conv2DTranspose(128, kernel_size=4, strides=2, padding="same"))  # 16x16
    model.add(BatchNormalization())
    model.add(LeakyReLU(0.2))

    model.add(Conv2DTranspose(64, kernel_size=4, strides=2, padding="same"))   # 32x32
    model.add(BatchNormalization())
    model.add(LeakyReLU(0.2))

    # 3 channels, tanh for [-1,1]
    model.add(Conv2D(3, kernel_size=3, padding="same", activation='tanh'))
    return model

# 3. Build the Discriminator (Conv Network):
def build_discriminator():
    model = Sequential(name="discriminator")
    model.add(Conv2D(64, kernel_size=3, strides=2, padding="same",
                     input_shape=(32, 32, 3)))
    model.add(LeakyReLU(0.2))
    model.add(Dropout(0.3))

    model.add(Conv2D(128, kernel_size=3, strides=2, padding="same"))
    model.add(LeakyReLU(0.2))
    model.add(Dropout(0.3))

    model.add(Flatten())
    model.add(Dense(1, activation='sigmoid'))
    return model
```

```python
# 4. Compile Models:
opt_D = Adam(0.0002, beta_1=0.5)
opt_G = Adam(0.0002, beta_1=0.5)

discriminator = build_discriminator()
discriminator.compile(loss='binary_crossentropy', optimizer=opt_D, metrics=['accuracy'])

generator = build_generator(latent_dim)
# Build combined (freeze D for this graph)
z = Input(shape=(latent_dim,))
fake_img = generator(z)
discriminator.trainable = False
validity = discriminator(fake_img)
dcgan = Model(z, validity, name="dcgan")
dcgan.compile(loss='binary_crossentropy', optimizer=opt_G)
discriminator.trainable = True  # restore for standalone D training
```

```python
# 5. Training Loop:
def train(epochs=5000, batch_size=64, save_interval=1000):
    half = batch_size // 2
    for epoch in range(epochs):
        # ---- Train Discriminator ----
        idx = np.random.randint(0, X_train.shape[0], half)
        real_imgs = X_train[idx]

        noise = np.random.normal(0, 1, (half, latent_dim))
        fake_imgs = generator.predict(noise, verbose=0)

        # Label smoothing for real
        real_y = 0.9 * np.ones((half, 1))
        fake_y = np.zeros((half, 1))

        d_loss_real = discriminator.train_on_batch(real_imgs, real_y)
        d_loss_fake = discriminator.train_on_batch(fake_imgs, fake_y)
        d_loss = 0.5 * (np.array(d_loss_real) + np.array(d_loss_fake))

        # ---- Train Generator (via combined model) ----
        noise = np.random.normal(0, 1, (batch_size, latent_dim))
        # Trick the discriminator: want them labeled as real (1)
        valid_y = np.ones((batch_size, 1))
        g_loss = dcgan.train_on_batch(noise, valid_y)

        if epoch % 100 == 0:
            print(f"{epoch:5d}  [D loss: {d_loss[0]:.4f}, acc: {100*d_loss[1]:.2f}%]  [G loss: {g_loss:.4f}]")

        if epoch % save_interval == 0:
            show_generated(epoch)
```

```python
# 6. Visualization Helper:
def show_generated(epoch):
    r, c = 5, 5
    noise = np.random.normal(0, 1, (r * c, latent_dim))
    gen_imgs = generator.predict(noise, verbose=0)
    gen_imgs = 0.5 * (gen_imgs + 1.0)  # [-1,1] -> [0,1]

    fig, axs = plt.subplots(r, c, figsize=(6, 6))
    count = 0
    for i in range(r):
        for j in range(c):
            axs[i, j].imshow(gen_imgs[count])
            axs[i, j].axis('off')
            count += 1
    plt.suptitle(f"DCGAN CIFAR-10 — Epoch {epoch}")
    plt.tight_layout()
    plt.show()
    plt.close()

# 7. Train DCGAN:
train(epochs=5000, batch_size=64, save_interval=1000)
```

By the end, the generator should take random noise (like a 100-dimensional vector) and produce images that look like real CIFAR-10 pictures, which means your network will learn the underlying distribution of the dataset without ever seeing labels.
