```python
# 1. Import Libraries and Dataset:
import numpy as np
import matplotlib.pyplot as plt

# 2. Build the Perceptron Model:
class Perceptron:
    def __init__(self, num_inputs, learning_rate=0.01):
        self.weights = np.random.rand(num_inputs + 1)
        self.learning_rate = learning_rate
    
    # Define the first linear layer 
    def linear(self, inputs):
        return np.dot(inputs, self.weights[1:]) + self.weights[0]
    
    # Define the Heaviside Step function.
    def Heaviside_step_fn(self, z):
        if z>=0:
            return 1
        else:
            return 0
        
    # Define the Prediction
    def predict(self, inputs):
        Z = self.linear(inputs)
        try:
            pred = []
            for z in Z:
                pred.append(self.Heaviside_step_fn(z))
        except:
            return self.Heaviside_step_fn(Z)
        return pred
    
    # Define the Loss function
    def loss(self, prediction, target):
        return target - prediction
    
    # Define training
    def train(self, inputs, target):
        prediction = self.predict(inputs)
        error = self.loss(prediction, target)
        self.weights[1:] += self.learning_rate * error * inputs
        self.weights[0]  += self.learning_rate * error
        
    # Fit the model
    def fit(self, X, y, num_epochs):
        for epoch in range(num_epochs):
            for inputs, target in zip(X, y):
                self.train(inputs, target)
    
    # Evaluating the model
    def evaluate(self, X, y):
        predictions = self.predict(X)
        return np.mean(predictions == y)

# 3. Training the Perceptron:
perceptron = Perceptron(num_features=2, learning_rate=0.1)
perceptron.train(X_train_scaled, y_train, epochs=5)
acc = perceptron.evaluate(X_test_scaled, y_test)

# 4. Evaluating the Model:
train_acc = perceptron.evaluate(X_train_scaled, y_train) * 100
test_acc = perceptron.evaluate(X_test_scaled, y_test) * 100

# 5. Decision Boundary:
def plot_decision_boundary(perceptron, X, y, title):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    x_vals = np.linspace(x_min, x_max, 200)

    w1, w2 = perceptron.weights[1:]
    b = perceptron.weights[0]

    if w2 != 0:  # avoid division by zero
        y_vals = -(w1/w2) * x_vals - (b/w2)
        plt.plot(x_vals, y_vals, 'k--', linewidth=2, label="Decision Boundary")

    # scatter plot of data
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.Paired)
    plt.title(title)
    plt.legend()
    plt.show()

# Plot for training set
plot_decision_boundary(perceptron, X_train_scaled, y_train, "Decision Boundary - Training Set")

# Plot for test set
plot_decision_boundary(perceptron, X_test_scaled, y_test, "Decision Boundary - Test Set")
```
